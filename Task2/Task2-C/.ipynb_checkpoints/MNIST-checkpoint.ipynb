```python
import tensorflow as tf
import numpy as np

tf.__version__
```

```python
( x_train, y_train_old ), ( x_test, y_test_old ) = tf.keras.datasets.mnist.load_data()

print ( x_train.shape )
print ( x_test.shape )
print ( y_train_old.shape )
print ( y_test_old.shape )
```

### Flatten each individual image so that we can feed it as a 1D vector into the neural network

```python
x_train = x_train.reshape ( -1, 28 * 28 )
x_test = x_test.reshape ( -1, 28 * 28 )
print ( x_train.shape )
print ( x_test.shape )
```

### Each y label is an integer, so convert to one hot labels. We are doing this because softmax is used, which will output a list of probabilities

```python
y_train = tf.one_hot ( tf.cast ( y_train_old, dtype = tf.int32 ), depth = 10 )
y_test = tf.one_hot ( tf.cast ( y_test_old, dtype = tf.int32 ), depth = 10 )
print ( y_train.shape )
print ( y_test.shape )
```

### gelu is a better relu :)

```python
model = tf.keras.Sequential ( [
    tf.keras.layers.InputLayer ( input_shape = ( 784, ) ),
    tf.keras.layers.Dense ( 256, activation = 'gelu' ),
    tf.keras.layers.Dense ( 128, activation = 'gelu' ),
    tf.keras.layers.Dense ( 10, activation = 'softmax' )
] )
```

```python
opt = tf.keras.optimizers.Adam ( learning_rate = 0.001 )
model.compile ( optimizer = opt, loss = "categorical_crossentropy" )
```

```python
model.fit ( x_train, y_train, epochs = 14, batch_size = 64, verbose = 2, validation_split = 0.2 )
```

### Final report!

```python
import numpy as np
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_probs = model.predict ( x_test )
y_pred = np.argmax ( y_probs, axis = 1 )
y_true = np.argmax ( y_test, axis = 1 )

acc = accuracy_score ( y_true, y_pred )
print ( "Test accuracy:", acc )

print ( "Confusion matrix: " )
print ( confusion_matrix ( y_true, y_pred ) )

print ( classification_report ( y_true, y_pred ) )
```
