```python
# Import libraries

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define constants/boundaries

GRID_SIZE = 5
START = (0, 0)
GOAL = (4, 4)
MAX_STEPS = 50
```

```python
# Actions

ACTIONS = {
    "UP": (-1, 0),
    "DOWN": (1, 0),
    "LEFT": (0, -1),
    "RIGHT": (0, 1)
}
```

```python
# Agent and grid definitions

class GridWorld:
    def __init__(self):
        self.reset()

    def reset(self):
        self.agent_pos = START
        self.steps = 0
        self.done = False
        self.path = [self.agent_pos]
        return self.agent_pos

    def step(self, action):
        if self.done:
            return self.agent_pos, 0, self.done

        dx, dy = ACTIONS[action]
        x, y = self.agent_pos
        nx, ny = x + dx, y + dy

        self.agent_pos = (nx, ny)

        self.steps += 1
        self.path.append(self.agent_pos)

        reward = -1
        if self.agent_pos == GOAL:
            reward = 10
            self.done = True
        elif self.steps >= MAX_STEPS:
            self.done = True

        return self.agent_pos, reward, self.done
```

```python
# Random policy

def random_policy(position, env):
    x, y = position
    valid = False
    while not valid:
        choice = np.random.choice ( [ "UP", "DOWN", "LEFT", "RIGHT" ] )
        dx, dy = ACTIONS[choice]
        nx, ny = x + dx, y + dy

        if 0 <= nx < GRID_SIZE and 0 <= ny < GRID_SIZE:
            valid = True
        else:
            valid = False

    return choice
```

```python
# Run 100 simulations

MAX_SIM = 100

env = GridWorld()

ep_reward = []
ep_len = []
success = 0

for i in range ( MAX_SIM ):
    state = env.reset()
    total_reward = 0
    cur_steps = 0
    
    while ( (not env.done) and (cur_steps < MAX_STEPS) ):
        action = random_policy(state, env)
        if action is None:
            break
        state, reward, done = env.step(action)
        total_reward += reward
        cur_steps += 1

    ep_reward.append ( total_reward )
    ep_len.append ( cur_steps )
    if state == GOAL:
        success += 1

print ( "Average reward: ", np.array ( ep_reward ).mean() )
print ( "Average number of steps: ", np.array ( ep_len ).mean() )
print ( "Number of successful runs: ", success )
```
